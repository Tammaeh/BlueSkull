```python

print("QUESTION 1: NUMPY - READ THE CSV FILE:")

df = pd.read_csv("/content/siyahamba_taxi_trips.csv")

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

df.head()
```

```python

# QUESTION 1.1: NUMPY - PASSENGER STATISTICS

# Convert the Passengers column into a NumPy array
passengers_array = df["Passengers"].to_numpy()

# Calculate the total number of passengers
total_passengers = np.sum(passengers_array)

# Calculate the average number of passengers per trip
mean_passengers = np.mean(passengers_array)

# Find the minimum number of passengers on a trip
minimum_passengers = np.min(passengers_array)

# Find the maximum number of passengers on a trip
maximum_passengers = np.max(passengers_array)

# Display the results
print("QUESTION 1.1: PASSENGER STATISTICS")
print("Total passengers:", total_passengers)
print("Average passengers per trip:", mean_passengers)
print("Minimum passengers per trip:", minimum_passengers)
print("Maximum passengers per trip:", maximum_passengers)
```

```python
# QUESTION 1.2: NUMPY - REVENUE SLICING

# Convert the TripRevenue_ZAR column into a NumPy array
revenue_array = df["TripRevenue_ZAR"].to_numpy()

# Display the first 15 revenue values
first_15_revenues = revenue_array[:15]

print("QUESTION 1.2: REVENUE SLICING")
print("First 15 revenue values:")
print(first_15_revenues)

# Select the last 50 revenue values
last_50_revenues = revenue_array[-50:]

# Select every fifth value from those 50 values
every_fifth_revenue = last_50_revenues[::5]

print("\nEvery fifth value from the last 50 revenue values:")
print(every_fifth_revenue)
```

```python
# QUESTION 1.3: NUMPY - DISTANCE STANDARD DEVIATION AND VARIANCE

# Convert the Distance_km column into a NumPy array
distance_array = df["Distance_km"].to_numpy()

# Calculate the standard deviation of trip distances
distance_std = np.std(distance_array)

# Calculate the variance of trip distances
distance_variance = np.var(distance_array)

# Display the results
print("QUESTION 1.3: DISTANCE VARIATION")
print("Standard deviation of trip distances:", distance_std)
print("Variance of trip distances:", distance_variance)
```

```python
# QUESTION 1.4: NUMPY - MANUAL REVENUE CALCULATION

# Select the number of passengers for the first 20 trips
passengers_first20 = df["Passengers"].head(20).to_numpy()

# Select the fare for the first 20 trips
fare_first20 = df["Fare_ZAR"].head(20).to_numpy()

# Calculate revenue by multiplying passengers by fare
manual_revenue = passengers_first20 * fare_first20

# Get the recorded revenue for the first 20 trips
recorded_revenue = df["TripRevenue_ZAR"].head(20).to_numpy()

# Display and compare the results
print("QUESTION 1.4: MANUAL REVENUE CALCULATION")

print("\nManually calculated revenue:")
print(manual_revenue)

print("\nRecorded revenue:")
print(recorded_revenue)

# Check whether the calculated and recorded values match
print("\nDo the values match?")
print(np.isclose(manual_revenue, recorded_revenue))
```

```python
# QUESTION 1.5: NUMPY - RESHAPE PASSENGER ARRAY

# Select the first 24 passenger values
passengers_first24 = df["Passengers"].head(24).to_numpy()

# Reshape the array into 6 rows and 4 columns
passengers_reshaped = passengers_first24.reshape(6, 4)

# Display the reshaped array
print("QUESTION 1.5: RESHAPED PASSENGER ARRAY")
print(passengers_reshaped)

# Display the shape of the new array
print("\nArray shape:", passengers_reshaped.shape)
```

```python
# QUESTION 2.1: PANDAS - DATASET INSPECTION

# Display the last 10 rows of the dataset
print("QUESTION 2.1: LAST 10 ROWS")
display(df.tail(10))

# Display the data type of each column
print("\nDATA TYPES")
print(df.dtypes)

# Display summary statistics for the dataset
print("\nSUMMARY STATISTICS")
display(df.describe(include="all"))
```

```python
# QUESTION 2.2: PANDAS - CHECK FOR MISSING VALUES

# Count the missing values in each column
missing_values = df.isnull().sum()

# Display the results
print("QUESTION 2.2: MISSING VALUES IN THE DATASET")
print(missing_values)
```

```python
# QUESTION 2.3: PANDAS - FILL MISSING VALUES

# Find the median rainfall value
rainfall_median = df["Rainfall_mm"].median()

# Fill missing rainfall values with the median
df["Rainfall_mm"] = df["Rainfall_mm"].fillna(rainfall_median)

# Find the median passenger rating
rating_median = df["PassengerRating"].median()

# Fill missing passenger ratings with the median
df["PassengerRating"] = df["PassengerRating"].fillna(rating_median)

# Check if any missing values remain in these columns
print("QUESTION 2.3: MISSING VALUES AFTER FILLING")
print(df[["Rainfall_mm", "PassengerRating"]].isnull().sum())
```

```python
# QUESTION 2.4: PANDAS - FILTER BARA PEAK-HOUR TRIPS

# Find trips where the rank is Bara
bara_trips = df[df["Rank"] == "Bara"]

# From those trips, select the ones during peak hours
bara_peak_trips = bara_trips[bara_trips["PeakHour"] == 1]

# Display the filtered trips
print("QUESTION 2.4: BARA TRIPS DURING PEAK HOURS")
display(bara_peak_trips)

# Count the number of trips found
print("Number of Bara peak-hour trips:", len(bara_peak_trips))
```

```python
# QUESTION 2.5: PANDAS - GROUP BY VEHICLE TYPE

# Group the data by vehicle type
vehicle_summary = df.groupby("VehicleType").agg(
    Total_Revenue=("TripRevenue_ZAR", "sum"),
    Average_Passengers=("Passengers", "mean")
)

# Display the results
print("QUESTION 2.5: REVENUE AND PASSENGERS BY VEHICLE TYPE")
display(vehicle_summary)
```

```python
# QUESTION 2.6: PANDAS - FUEL COSTS AND LONGEST TRIPS

# Estimate fuel costs as 25% of the trip revenue
df["FuelCost_Est"] = df["TripRevenue_ZAR"] * 0.25

# Display the estimated fuel costs
print("QUESTION 2.6: ESTIMATED FUEL COSTS")
display(df[["TripRevenue_ZAR", "FuelCost_Est"]].head())

# Find the five trips with the longest distances
longest_trips = df.nlargest(5, "Distance_km")

# Display the five longest trips
print("\nFIVE LONGEST TRIPS")
display(longest_trips)
```

```python
# QUESTION 3: MATPLOTLIB - CHECK COLUMN NAMES

print("QUESTION 3: CHECK COLUMN NAMES")
print(df.columns.tolist())
```

```python
# QUESTION 3: MATPLOTLIB - CALCULATE WEEKLY REVENUE

# Convert the Date column into a date format
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

# Group the trips by week and calculate total revenue
weekly_revenue = df.groupby(
    pd.Grouper(key="Date", freq="W")
)["TripRevenue_ZAR"].sum()

# Display the weekly revenue
print("QUESTION 3: TOTAL REVENUE PER WEEK")
display(weekly_revenue)
```

```python
# QUESTION 3: MATPLOTLIB - WEEKLY REVENUE GRAPH

# Create the line graph
plt.figure(figsize=(12, 6))

plt.plot(
    weekly_revenue.index,
    weekly_revenue.values,
    marker="o",
    linestyle="-"
)

# Add a title and labels
plt.title("Siyahamba Taxi Association: Weekly Revenue")
plt.xlabel("Week")
plt.ylabel("Total Revenue (ZAR)")

# Rotate the dates to make them easier to read
plt.xticks(rotation=45)

# Add a grid to the graph
plt.grid(True)

# Adjust the layout
plt.tight_layout()

# Save the graph as an image
plt.savefig("weekly_revenue.png", dpi=300, bbox_inches="tight")

# Display the graph
plt.show()

print("Weekly revenue graph saved successfully!")
```

```python
# QUESTION 4.1: DATA VISUALISATION - PASSENGERS BY VEHICLE TYPE

# Calculate the total passengers for each vehicle type
passengers_by_vehicle = df.groupby("VehicleType")["Passengers"].sum()

# Create a bar chart
plt.figure(figsize=(8, 5))
passengers_by_vehicle.plot(kind="bar")

# Add a title and labels
plt.title("Total Passengers by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Total Passengers")

# Make the chart easier to read
plt.xticks(rotation=45)
plt.tight_layout()

# Display the chart
plt.show()
```

```python
# QUESTION 4.2: DATA VISUALISATION - REVENUE BY RANK

# Calculate total revenue for each rank
revenue_by_rank = df.groupby("Rank")["TripRevenue_ZAR"].sum()

# Create a bar chart
plt.figure(figsize=(8, 5))
revenue_by_rank.plot(kind="bar")

# Add a title and labels
plt.title("Total Trip Revenue by Taxi Rank")
plt.xlabel("Taxi Rank")
plt.ylabel("Total Revenue (ZAR)")

# Make the chart easier to read
```

```python
# QUESTION 4.3: DATA VISUALISATION - PASSENGER RATINGS

# Calculate the average passenger rating for each vehicle type
rating_by_vehicle = df.groupby("VehicleType")["PassengerRating"].mean()

# Create a bar chart
plt.figure(figsize=(8, 5))
rating_by_vehicle.plot(kind="bar")

# Add a title and labels
plt.title("Average Passenger Rating by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Average Passenger Rating")

# Make the chart easier to read
plt.xticks(rotation=45)
plt.tight_layout()

# Display the chart
plt.show()
```

```python
# QUESTION 4.4: DATA VISUALISATION - DISTANCE BY VEHICLE TYPE

# Calculate the total distance for each vehicle type
distance_by_vehicle = df.groupby("VehicleType")["Distance_km"].sum()

# Create a bar chart
plt.figure(figsize=(8, 5))
distance_by_vehicle.plot(kind="bar")

# Add a title and labels
plt.title("Total Distance Travelled by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Total Distance (km)")

# Make the chart easier to read
plt.xticks(rotation=45)
plt.tight_layout()

# Display the chart
plt.show()
```

```python
# QUESTION 4.5: DATA VISUALISATION - REVENUE VS DISTANCE

# Create a scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Distance_km"],
    df["TripRevenue_ZAR"]
)

# Add a title and labels
plt.title("Trip Revenue Compared to Distance Travelled")
plt.xlabel("Distance Travelled (km)")
plt.ylabel("Trip Revenue (ZAR)")

# Display the chart
plt.tight_layout()
plt.show()
```

```python
# QUESTION 5: SUMMARY - IMPORTANT DATA RESULTS

print("QUESTION 5: SUMMARY OF DATA ANALYSIS")

# Calculate total revenue
total_revenue = df["TripRevenue_ZAR"].sum()
print("Total Revenue (ZAR):", round(total_revenue, 2))

# Calculate total passengers
total_passengers = df["Passengers"].sum()
print("Total Passengers:", total_passengers)

# Find the vehicle type with the highest revenue
revenue_by_vehicle = df.groupby("VehicleType")["TripRevenue_ZAR"].sum()
print("Vehicle Type with Highest Revenue:", revenue_by_vehicle.idxmax())

# Find the taxi rank with the highest revenue
revenue_by_rank = df.groupby("Rank")["TripRevenue_ZAR"].sum()
print("Taxi Rank with Highest Revenue:", revenue_by_rank.idxmax())

# Calculate the average passenger rating
average_rating = df["PassengerRating"].mean()
print("Average Passenger Rating:", round(average_rating, 2))

# Calculate the average trip distance
average_distance = df["Distance_km"].mean()
print("Average Trip Distance (km):", round(average_distance, 2))
```
